```python
import sys
import os
from pathlib import Path

print("=" * 60)
print("1. CWD actual:")
print(f"   {os.getcwd()}")
print(f"   Nombre: {Path.cwd().name}")

print("\n2. Contenido del CWD:")
for f in sorted(os.listdir('.'))[:20]:
    print(f"   {f}")

print("\n3. project_root calculado:")
project_root = Path.cwd().parent
print(f"   {project_root}")
print(f"   ¿Existe?: {project_root.exists()}")

print("\n4. src path:")
src_path = project_root / 'src'
print(f"   {src_path}")
print(f"   ¿Existe?: {src_path.exists()}")

if src_path.exists():
    print(f"\n5. Contenido de src/:")
    for f in sorted(os.listdir(src_path)):
        print(f"   {f}")
    
    pipelines_path = src_path / 'pipelines'
    print(f"\n6. ¿Existe src/pipelines?: {pipelines_path.exists()}")
    if pipelines_path.exists():
        print(f"   Contenido:")
        for f in sorted(os.listdir(pipelines_path)):
            print(f"      {f}")

print("\n7. sys.path después de insertar:")
sys.path.insert(0, str(src_path))
for p in sys.path[:5]:
    print(f"   {p}")
```
